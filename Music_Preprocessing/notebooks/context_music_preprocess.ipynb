```python
from google.colab import files
from pathlib import Path

raw_dir = Path('data/raw_music')
raw_dir.mkdir(parents=True, exist_ok=True)
uploaded = files.upload()
for name, content in uploaded.items():
    (raw_dir / name).write_bytes(content)
print('Uploaded files:', [p.name for p in raw_dir.iterdir()])
```

```python
import json
from pathlib import Path

Path('config').mkdir(exist_ok=True)
prompts = {
    'forest': 'music that feels suitable for a forest or woodland environment, with trees, leaves, birds, shade, natural calm, and a strong green-space atmosphere',
    'water': 'music that feels suitable for blue-space environments such as rivers, lakes, beaches, ocean coastlines, streams, canals, waves, and waterside walking',
    'park': 'music that feels suitable for an urban public park, garden, green recreation area, walking path, grass, leisure, and relaxed outdoor activity',
    'urban': 'music that feels suitable for a dense built environment, city streets, buildings, commercial areas, shops, cafes, and everyday urban activity',
    'road': 'music that feels suitable for roads, transport corridors, traffic movement, cycling, commuting, driving, and forward motion',
    'campus': 'music that feels suitable for a university or school campus, study walking, courtyards, academic buildings, and student life'
}
Path('config/environment_prompts.json').write_text(json.dumps(prompts, indent=2), encoding='utf-8')
```
